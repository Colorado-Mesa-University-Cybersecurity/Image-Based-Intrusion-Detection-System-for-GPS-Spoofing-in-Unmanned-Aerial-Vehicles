```python
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from supervised.automl import AutoML
import warnings
from tqdm import TqdmWarning
from sklearn.model_selection import train_test_split

# Removes annoying tqdm warnings
warnings.filterwarnings("ignore", category=TqdmWarning)

# This warning is related to autoML which is has not fixed the memory leak issue on Windows with MKL.
warnings.filterwarnings(
    "ignore",
    message="MiniBatchKMeans is known to have a memory leak on Windows with MKL, when there are less chunks than available threads"
)
```

```python
# Constants
RANDOM_STATE = 1082
TEST_SIZE = 0.2
MODEL_TIME_LIMIT = 60 * 60 # 1 hour
Total_Time_Limit = 60 * 60 * 11 # 11 hours
```

```python
# Load the dataset
try:
    df = pd.read_csv("../../Datasets/csv/spoofing-merged-gps-only.csv")
except FileNotFoundError:
    print("Error: The file '../../Datasets/csv/spoofing-merged-gps-only.csv' was not found.")
    print("Please make sure the path is correct.")
    exit()
```

```python
# Data preprocessing

# Clean up column names
df.columns = df.columns.str.replace('[\\[\\]<>]', '_', regex=True)

# Drop timestamp column, keep 'label' as target variable
drop_cols = ['timestamp']
X = df.drop(columns=drop_cols + ['label'])
y = df['label']

# Encode label
if y.dtype == object or y.dtype.name == 'category':
    from sklearn.preprocessing import LabelEncoder
    le = LabelEncoder()
    y = le.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y
)
```

```python
# Train the AutoML models
automl = AutoML(mode="Compete", random_state=RANDOM_STATE, results_path="competeResults", model_time_limit=MODEL_TIME_LIMIT, explain_level=1, total_time_limit=Total_Time_Limit)
automl.fit(X_train, y_train)
```

```python
y_pred = automl.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {acc:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
