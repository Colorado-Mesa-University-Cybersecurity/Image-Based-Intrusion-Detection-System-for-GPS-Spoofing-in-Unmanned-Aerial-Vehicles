```python
import csv
import os

# Input and output file paths
file_pairs = [
    ("../../../UAVATTACK_Results.csv", "../UAVATTACK_Results_filtered.csv"),
    ("../../../2D_Results.csv", "../2D_Results_filtered.csv"),
]

# Columns to keep
keep_columns = ["model", "dataset", "family", "mcc", "balanced_accuracy", "f1_macro",
                "g_mean", "accuracy", "majority_baseline_accuracy",
                "accuracy_lift_over_majority", "f1_weighted", "is_degenerate",
                "recall_c0", "recall_c1", "recall_c2", "recall_c3",
                "prior_shift_tv", "infer_batch_us_per_row",
                "infer_online_p50_us", "infer_online_p95_us",
                "train_time_sec", "num_epochs"]

for input_file, output_file in file_pairs:
    if not os.path.isfile(input_file):
        print(f"Skipping {input_file}: file not found")
        continue

    with open(input_file, "r", newline='', encoding="utf-8") as infile, \
         open(output_file, "w", newline='', encoding="utf-8") as outfile:
        reader = csv.DictReader(infile)
        writer = csv.DictWriter(outfile, fieldnames=keep_columns)
        writer.writeheader()
        for row in reader:
            # Only select the desired columns
            filtered_row = {col: row.get(col, "") for col in keep_columns}
            writer.writerow(filtered_row)

    print(f"Wrote {output_file}")
```
