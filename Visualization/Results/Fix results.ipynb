    Reads the results CSV file that has parsing errors due to unquoted commas
    in the 'parameters' column and writes a corrected version.

```python
import csv
import os

def clean_csv_file(original_path, corrected_path):
    try:
        with open(original_path, 'r', encoding='utf-8') as infile, \
             open(corrected_path, 'w', newline='', encoding='utf-8') as outfile:

            # Use the csv module for robust handling
            writer = csv.writer(outfile)

            # Read the header to determine the expected number of columns
            header_line = infile.readline()
            header = header_line.strip().split(',')
            num_expected_columns = len(header)
            writer.writerow(header)

            print(f"Header found with {num_expected_columns} columns: {header}")

            # Process the rest of the file
            for line_num, line in enumerate(infile, start=2):
                fields = line.strip().split(',')

                if len(fields) > num_expected_columns:
                    print(f"Line {line_num}: Found {len(fields)} fields. Fixing")
                    params_field = ",".join(fields[5:-1])
                    corrected_row = fields[:5] + [params_field] + fields[-1:]

                    writer.writerow(corrected_row)
                else:
                    writer.writerow(fields)

        print(f"\nSuccessfully created corrected file at: {corrected_path}")

    except FileNotFoundError:
        print(f"Error: The file '{original_path}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == '__main__':
    input_csv_path = os.path.join('..', '..', '2D_Results.csv')
    output_csv_path = os.path.join('..', '..', '2D_Results_full.csv')
    clean_csv_file(input_csv_path, output_csv_path)
    input_csv_path = os.path.join('..', '..', 'UAVATTACK_Results.csv')
    output_csv_path = os.path.join('..', '..', 'UAVATTACK_Results_full.csv')
    clean_csv_file(input_csv_path, output_csv_path)
```
