```python
import csv

# Input and output file paths
input_file = "../../UAVATTACK_Results.csv"
output_file = "UAVATTACK_Results_filtered.csv"

# Columns to keep
keep_columns = ["model", "dataset", "accuracy", "train_time_sec", "num_epochs"]

with open(input_file, "r", newline='', encoding="utf-8") as infile, \
     open(output_file, "w", newline='', encoding="utf-8") as outfile:
    reader = csv.DictReader(infile)
    writer = csv.DictWriter(outfile, fieldnames=keep_columns)
    writer.writeheader()
    for row in reader:
        # Only select the desired columns
        filtered_row = {col: row[col] for col in keep_columns}
        writer.writerow(filtered_row)
```
