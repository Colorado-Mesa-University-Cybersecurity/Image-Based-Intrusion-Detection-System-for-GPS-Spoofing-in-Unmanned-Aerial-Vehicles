```python
from gps_ids import protocol, training
```

```python
feature_cols_1 = ['TCD', 'PC', 'CN0', 'EC', 'PRN', 'LC', 'DO', 'PD' ,'CP', 'PIP']
feature_cols_2 = ['TOW', 'PD', 'PRN', 'RX', 'TCD', 'DO', 'CP', 'CN0', 'PC', 'EC']

EXP1 = protocol.real_heldout("2d", features=feature_cols_1)
EXP2 = protocol.real_heldout("2d", features=feature_cols_2)
```

```python
for base in ("XGBoost", "MLP Classifier", "Logistic Regression", "Random Forest"):
    training.run(f"{base} (Explainability Features)", EXP1)
    training.run(f"{base} (Xgboost Features)", EXP2)
```

```python
# Seed sweep: the remaining seeds in training.SEEDS, appended to the results CSV.
SWEEP_SEEDS = [s for s in training.SEEDS if s != protocol.SEED]

for base in ("XGBoost", "MLP Classifier", "Logistic Regression", "Random Forest"):
    training.run_seeds(
        f"{base} (Explainability Features)",
        lambda s: protocol.real_heldout("2d", features=feature_cols_1, seed=s),
        seeds=SWEEP_SEEDS,
    )
    training.run_seeds(
        f"{base} (Xgboost Features)",
        lambda s: protocol.real_heldout("2d", features=feature_cols_2, seed=s),
        seeds=SWEEP_SEEDS,
    )
```
