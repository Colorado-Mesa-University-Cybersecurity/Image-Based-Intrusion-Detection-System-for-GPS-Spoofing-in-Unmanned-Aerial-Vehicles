```python
from gps_ids import protocol, training
```

```python
# Feature sets
feature_sets = [
    {
        "name": "LSTM Classifier (Explainability Features)",
        "features": [
            'vdop', 'ax', 'hdop', 'cog_rad', 'jamming_indicator',
            'terrain_alt_valid', 'dist_bottom_valid', 'terrain_alt',
            'lat_x', 'lat_y'
        ]
    },
    {
        "name": "LSTM Classifier (Xgboost Features)",
        "features": [
            'lat_y', 'vel_e_m_s', 'vel_n_m_s', 'c_variance_rad', 'hdop',
            's_variance_m_s', 'q_0_', 'vel_m_s', 'cog_rad', 'q_3_'
        ]
    }
]
```

```python
for fs in feature_sets:
    EXP = protocol.real_heldout("uav", features=fs["features"])
    training.run(fs["name"], EXP)
```

```python
# Seed sweep: the remaining seeds in training.SEEDS, appended to the results CSV.
SWEEP_SEEDS = [s for s in training.SEEDS if s != protocol.SEED]

for fs in feature_sets:
    training.run_seeds(
        fs["name"],
        lambda s, fs=fs: protocol.real_heldout("uav", features=fs["features"], seed=s),
        seeds=SWEEP_SEEDS,
    )
```
