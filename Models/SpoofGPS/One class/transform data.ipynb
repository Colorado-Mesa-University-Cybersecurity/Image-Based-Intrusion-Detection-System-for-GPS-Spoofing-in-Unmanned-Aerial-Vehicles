```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Load the dataset
try:
    df = pd.read_csv("../../../Datasets/csv/spoofing-merged-gps-only.csv")
except FileNotFoundError:
    print("Error: The file '../../Datasets/csv/spoofing-merged-gps-only.csv' was not found.")
    exit()
```

```python
# Clean up column
df.columns = df.columns.str.replace('[\\[\\]<>]', '_', regex=True)
```

```python
# Split features and label
X = df.drop(columns=['timestamp', 'label'])
y = df['label']

# Encode labels
from sklearn.preprocessing import LabelEncoder
if y.dtype == object or y.dtype.name == 'category':
    le = LabelEncoder()
    y = le.fit_transform(y)
```

```python
# Select only benign samples (0 = benign, 1 = malicious)
benign_mask = y == 0
X_benign = X[benign_mask]
y_benign = y[benign_mask]

# Split benign samples into training and validation sets
X_train, X_benign_val, y_train, y_benign_val = train_test_split(
    X_benign, y_benign, test_size=0.2, random_state=1082
)

# Test set contains BOTH benign and malicious samples
X_test = X
y_test = y

print(f"Training set size (benign only): {X_train.shape[0]}")
print(f"Test set size (benign + malicious): {X_test.shape[0]}")
```
