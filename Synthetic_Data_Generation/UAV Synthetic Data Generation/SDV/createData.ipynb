```python
from sdv.single_table import GaussianCopulaSynthesizer, CTGANSynthesizer, TVAESynthesizer
```

```python
# Load synthesizers
gc_synth = GaussianCopulaSynthesizer.load('spoofing-GaussianCopula-synthesizer.pkl')
ctgan_synth = CTGANSynthesizer.load('spoofing-CTGAN-synthesizer.pkl')
tvae_synth = TVAESynthesizer.load('spoofing-TVAE-synthesizer.pkl')
```

```python
# Sample sizes
sample_sizes = [10_000, 100_000, 1_000_000]
models = {
    'gc': gc_synth,
    'ctgan': ctgan_synth,
    'tvae': tvae_synth
}
output_dir = '../../../Datasets/csv/UAVATTACK/'
```

```python
# Generate and save datasets
for model_name, synth in models.items():
    for n in sample_sizes:
        synth.reset_sampling()
        df = synth.sample(n)
        df.to_csv(f'{output_dir}spoofing_{model_name}_{n}.csv', index=False)
```
